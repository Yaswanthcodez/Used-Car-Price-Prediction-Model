```python
num_cols=["Year","Kilometer","Engine","Power","Power rpm","Torque","Torque rpm","Length","Width","Height","Seating Capacity","Fuel Tank Capacity"]
cat_cols=["Make","Fuel Type","Transmission","Location","Color","Owner","Seller Type","Drivetrain"]
```

```python
x=df.drop(columns=["Price"])
y=df["Price"]
```

```python
print(len(num_cols))
print(len(cat_cols))
print(x.shape)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler,OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.impute import SimpleImputer

num_pipeline=Pipeline([("imputer",SimpleImputer(strategy="median")),("scaler",StandardScaler())])
cat_pipeline=Pipeline([("imputer",SimpleImputer(strategy="most_frequent")),("encoder",OneHotEncoder(handle_unknown="ignore"))])
```

```python
preprocessor=ColumnTransformer([("num",num_pipeline,num_cols),("cat",cat_pipeline,cat_cols)])
```

```python
from sklearn.ensemble import RandomForestRegressor
rf=RandomForestRegressor()
```

```python
pipe=Pipeline([("preprocessor",preprocessor),("model",rf)])
```

```python
pipe.fit(x_train,y_train)
```

```python
y_pred=pipe.predict(x_test)
print("MAE:", mean_absolute_error(y_test,y_pred))
print("RMSE:", mean_squared_error(y_test,y_pred)**0.5)

from sklearn.model_selection import cross_val_score
cv_scores = cross_val_score(pipe,x,y,cv=5,scoring="r2",n_jobs=-1)
print("Cross-validation R2:", cv_scores)
print("Mean cross-validation R2:", cv_scores.mean())
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
```

```python
print(pipe.score(x_train,y_train))
print(pipe.score(x_test,y_test))
```

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
```

```python
pipe_lr=Pipeline([("preprocessor",preprocessor),("model",lr)])
```

```python
pipe_lr.fit(x_train,y_train)
```

```python
print(pipe_lr.score(x_train,y_train))
print(pipe_lr.score(x_test,y_test))
```

```python
from sklearn.tree import DecisionTreeRegressor
dt=DecisionTreeRegressor()
```

```python
pipe_dt=Pipeline([("preprocessor",preprocessor),("model",dt)])
```

```python
pipe_dt.fit(x_train,y_train)
```

```python
print(pipe_dt.score(x_train,y_train))
print(pipe_dt.score(x_test,y_test))
```
